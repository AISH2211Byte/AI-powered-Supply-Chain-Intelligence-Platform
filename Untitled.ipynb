```python
y = df["Late_delivery_risk"]

X = df.drop(columns=["Late_delivery_risk"]).copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
leakage_cols = [
    "Days for shipping (real)",
    "Delivery Status",
    "shipping date (DateOrders)"
]

X = X.drop(
    columns=[col for col in leakage_cols if col in X.columns]
)
```

```python
print("Number of features:", X.shape[1])

for col in X.columns:
    print(col)
```

```python
id_cols = [
    "Order Id",
    "Order Item Id",
    "Customer Id",
    "Product Card Id",
    "Product Category Id",
    "Category Id",
    "Department Id"
]

X = X.drop(
    columns=[col for col in id_cols if col in X.columns]
)

print("Features remaining:", X.shape[1])
```

```python
categorical_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()
numerical_cols = X.select_dtypes(include=["number"]).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)

print("\nNo. categorical:", len(categorical_cols))
print("No. numerical:", len(numerical_cols))
```

```python
drop_cols = [
    "Order Customer Id",
    "Order Item Cardprod Id",
    "Customer Street",
    "Customer City",
    "Customer Country",
    "Customer State",
    "Order Status"
]

X = X.drop(
    columns=[col for col in drop_cols if col in X.columns]
)

print("Final number of features:", X.shape[1])
```

```python
categorical_cols = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

numerical_cols = X.select_dtypes(
    include=["number"]
).columns.tolist()

print("Categorical:", len(categorical_cols))
print("Numerical:", len(numerical_cols))
print("Total:", len(categorical_cols) + len(numerical_cols))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

# Numerical preprocessing
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])

# Combine both
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_cols),
        ("cat", categorical_transformer, categorical_cols)
    ]
)

print("Preprocessor created successfully.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        random_state=42,
        max_iter=1000
    ))
])
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:,1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("ROC AUC  :", roc_auc_score(y_test, y_prob))
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)
```
